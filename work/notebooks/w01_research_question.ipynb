## 3. Quick look at the data (2-3 real numbers)
The starter dataset contains approximately 30,000 pages, providing enough examples for exploration. The correlation between search volume and impressions is very weak (around 0.001 in my experiment), suggesting that search volume alone is not a reliable indicator of page traffic. The comparison of median word count across trend groups also suggests that longer content alone does not explain whether a page grows or declines. These observations support exploring multiple features together rather than relying on a single metric.

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Rows:", len(df))
print("Columns:", len(df.columns))

corr = df["search_volume"].corr(df["impressions_90d"])
print("\nCorrelation between search volume and impressions:", round(corr, 3))

print("\nMedian word count by trend:")
print(df.groupby("trend_direction")["word_count"].median())
```
